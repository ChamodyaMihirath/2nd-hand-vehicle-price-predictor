```python
df["Price"] = (
    df["Price"]
    .astype(str)
    .str.replace(r"[^0-9]", "", regex=True)
)

df["Price"] = pd.to_numeric(df["Price"], errors="coerce")
```

```python
df.head()
```

```python
df["Mileage"] = (
    df["Mileage"]
    .astype(str)
    .str.replace(r"[^0-9]", "", regex=True)
)

df["Mileage"] = pd.to_numeric(df["Mileage"], errors="coerce")
```

```python
df.head()
```

```python
df["Engine capacity"] = (
    df["Engine capacity"]
    .astype(str)
    .str.replace(r"[^0-9]", "", regex=True)
    .str.strip()
)

df["Engine capacity"] = pd.to_numeric(
    df["Engine capacity"],
    errors="coerce"
)
```
